```python
from ml.tcn.predictor import Predictor
import pandas as pd
import numpy as np
```

```python
import numpy as np

def collect_data(path):
    with np.load(path) as data:
        X = data['X']
        y = data['y']
        return X, y
```

```python
train_X, train_y = collect_data('robot_data_train.npz')
test_X, test_y = collect_data('robot_data_test.npz')
```

```python
test_y.sum() / test_y.shape[0], train_y.sum() / train_y.shape[0]
```

```python
INPUT_DIM = train_X.shape[2]
SEQ_LEN = train_X.shape[1]
INPUT_DIM, SEQ_LEN
```

```python
predictor = Predictor(
    input_dim=INPUT_DIM,
    num_classes=2,
    tcn_channels=tuple([52, 52]),
    dropout=0.9,
    epochs=20,
)
```

```python
from sklearn.metrics import f1_score


def score(X, y) -> float:
    predictions = predictor.predict(X)
    f1 = f1_score(predictions, y, average='weighted')
    return f1


def get_scores() -> None:
    train_score = score(train_X, train_y)
    test_score = score(test_X, test_y)
    print(f'Train f1-score: {train_score:.4f}\t Test f1-score: {test_score:.4f}')
```

```python
predictor.fit(train_X, train_y, eval_set=(test_X, test_y), verbose=True)
```

```python
get_scores()
```
